```python
#### import json
from datetime import datetime

class BudgetTracker:
    # Main class for the Budget Tracker application.
    # Demonstrates: classes, __init__, methods, instance variables,
    # try/except, conditionals, dictionaries, and f-strings.

    def __init__(self, name):
        # Constructor - initializes all variables when object is created.
        # self. means these variables belong to this specific object.
        self.name = name # store user's name
        self.date = datetime.now().strftime("%Y-%m-%d %H:%M") # store current date and time

        # income
        self.monthly_income = 0.0 # main monthly income
        self.other_income = 0.0 # additional income sources

        # expense categories
        self.food = 0.0 # food and groceries
        self.transportation = 0.0 # transport costs
        self.entertainment = 0.0 # entertainment spending
        self.housing = 0.0 # rent or mortgage
        self.utilities = 0.0 # electricity, water, etc.
        self.other_expenses = 0.0 # any other expenses

        # savings
        self.savings = 0.0 # regular monthly savings
        self.retirement = 0.0 # retirement fund (401K)
        self.investments = 0.0 # investment contributions

    def get_input(self, prompt):
        # Safe input method using try/except.
        # Keeps asking until user enters a valid positive number.
        # Demonstrates: try/except, while loop, ValueError handling.
        while True:
            try:
                value = float(input(prompt)) # try to convert input to a number
                if value < 0: # check if number is negative
                    print("Please enter a positive number.")
                    continue # go back to the top of the loop
                return value # input is valid, return it
            except ValueError: # catches error if user types letters
                print("Invalid input. Please enter a number.")

    def enter_income(self):
        # Method to collect income data from user
        print("\n Income ") # section header for income
        self.monthly_income = self.get_input("  Monthly income (after tax): $") # ask for main income
        self.other_income = self.get_input("  Other income (side jobs, etc.): $") # ask for extra income

    def enter_expenses(self):
        # Method to collect expense data by category.
        # Categories: food, transportation, entertainment, housing, utilities, other expenses
        print("\n Monthly Expenses by Category ") # section header for expenses
        self.food = self.get_input("  Food & groceries: $") # ask for food costs
        self.transportation = self.get_input("  Transportation: $") # ask for transport costs
        self.entertainment = self.get_input("  Entertainment: $") # ask for entertainment costs
        self.housing = self.get_input("  Housing (rent/mortgage): $") # ask for housing costs
        self.utilities = self.get_input("  Utilities: $") # ask for utility bills
        self.other_expenses = self.get_input("  Other expenses: $") # ask for any other costs

    def enter_savings(self):
        # Method to collect savings and investment data
        print("\n Savings & Investments ") # section header for savings
        self.savings = self.get_input("  Monthly savings: $") # regular saving amount
        self.retirement = self.get_input("  Retirement (401k): $") # retirement contributions
        self.investments = self.get_input("  Investments: $") # investment contributions

    def total_income(self):
        # Returns total income (basic math operation)
        return self.monthly_income + self.other_income # add all income sources together

    def total_expenses(self):
        # Returns sum of all expense categories (basic math operation)
        return (self.food + self.transportation + self.entertainment +
                self.housing + self.utilities + self.other_expenses) # sum all expense categories

    def total_savings(self):
        # Returns total savings (basic math operation)
        return self.savings + self.retirement + self.investments # sum all savings

    def net_balance(self):
        # Returns remaining balance after expenses and savings.
        # Demonstrates: basic math operations.
        return self.total_income() - self.total_expenses() - self.total_savings() # money left over

    def expense_ratio(self):
        # Calculates what percentage of income goes to expenses.
        # Demonstrates: conditional logic to avoid division by zero.
        if self.total_income() == 0: # avoid division by zero error
            return 0
        return (self.total_expenses() / self.total_income()) * 100 # expenses as % of income

    def check_50_30_20(self):
        # Calculates the 50/30/20 budgeting rule limits.
        # Demonstrates: basic math, returning multiple values.
        # 50% for needs, 30% for wants, 20% for savings.
        needs_limit = self.total_income() * 0.50 # 50% of income for needs
        wants_limit = self.total_income() * 0.30 # 30% of income for wants
        savings_limit = self.total_income() * 0.20 # 20% of income for savings
        return needs_limit, wants_limit, savings_limit # return all 3 numbers together

    def generate_report(self):
        # Generates a formatted budget report as a string.
        # Demonstrates: f-strings, conditional logic, string formatting.
        needs_limit, wants_limit, savings_limit = self.check_50_30_20() # get all 3 numbers at once

        # build report string using f-strings
        report = ""
        report = report + "-" * 50 + "\n"
        report = report + f"BUDGET REPORT - {self.name}\n"
        report = report + f"Date: {self.date}\n"
        report = report + "-" * 50 + "\n"
        report = report + "\n Income\n"
        report = report + f"  Monthly income:    ${self.monthly_income:.2f}\n"
        report = report + f"  Other income:      ${self.other_income:.2f}\n"
        report = report + f"  TOTAL INCOME:      ${self.total_income():.2f}\n"
        report = report + "\n Expenses by Category\n"
        report = report + f"  Food & groceries:  ${self.food:.2f}\n"
        report = report + f"  Transportation:    ${self.transportation:.2f}\n"
        report = report + f"  Entertainment:     ${self.entertainment:.2f}\n"
        report = report + f"  Housing:           ${self.housing:.2f}\n"
        report = report + f"  Utilities:         ${self.utilities:.2f}\n"
        report = report + f"  Other:             ${self.other_expenses:.2f}\n"
        report = report + f"  TOTAL EXPENSES:    ${self.total_expenses():.2f}\n"
        report = report + "\n Savings\n"
        report = report + f"  Monthly savings:   ${self.savings:.2f}\n"
        report = report + f"  Retirement:        ${self.retirement:.2f}\n"
        report = report + f"  Investments:       ${self.investments:.2f}\n"
        report = report + f"  TOTAL SAVINGS:     ${self.total_savings():.2f}\n"
        report = report + "\n Summary\n"
        report = report + f"  Net balance:       ${self.net_balance():.2f}\n"
        report = report + f"  Expense ratio:     {self.expense_ratio():.1f}%\n"
        report = report + "\n 50/30/20 Rule\n"
        report = report + f"  50% needs limit:   ${needs_limit:.2f}  (your expenses: ${self.total_expenses():.2f})\n"
        report = report + f"  30% wants limit:   ${wants_limit:.2f}  (your savings: ${self.total_savings():.2f})\n"
        report = report + f"  20% savings limit: ${savings_limit:.2f}\n"

        # Alerts
        alerts = [] # start with empty list
        if self.net_balance() < 0: # check if spending more than earning
            alerts.append("Your expenses exceed your income!")
        if self.total_expenses() > needs_limit: # check if expenses exceed 50% limit
            alerts.append("Expenses exceed 50% rule limit.")
        if self.total_savings() < savings_limit: # check if saving below 20% limit
            alerts.append("Savings below 20% rule recommendation.")
        if not alerts: # if list is still empty = no problem found
            alerts.append("Your budget looks healthy!")

        report = report + "\n Alerts\n"
        for alert in alerts: # go through each alert one by one
            report = report + " " + alert + "\n"

        return report # return the complete report string

    def to_dict(self):
        # Converts object data to a dictionary for saving to JSON file.
        # Demonstrates: dictionaries, nested dictionaries.
        return {
            "name": self.name,
            "date": self.date,
            "income": {
                "monthly": self.monthly_income,
                "other": self.other_income,
                "total": self.total_income()
            },
            "expenses": {
                "food": self.food,
                "transportation": self.transportation,
                "entertainment": self.entertainment,
                "housing": self.housing,
                "utilities": self.utilities,
                "other": self.other_expenses,
                "total": self.total_expenses()
            },
            "savings": {
                "savings": self.savings,
                "investments": self.investments,
                "retirement": self.retirement,
                "total": self.total_savings()
            },
            "summary": {
                "net_balance": self.net_balance(),
                "expense_ratio": self.expense_ratio()
            }
        }

# FILE I/O
def save_data(history, filename="budget_history.json"):
    # Saves all budget sessions to JSON format (file handling)
    # Saves all budget sessions to a JSON file (File I/O)
    try:
        with open(filename, "w") as f: # open file for writing
            json.dump(history, f, indent=4) # convert Python data to JSON and save
            print("Data saved to budget_history.json") # confirm successful save
    except: # catch any error if file cannot be saved
        print("Error saving data.") # print error message to user

def load_data(filename="budget_history.json"):
    # Loads previous budget sessions from JSON file (File I/O)
    try:
        with open(filename, "r") as f: # open file for reading
            return json.load(f) # convert JSON back to Python data
    except: # no file found, start with empty list
        return [] # return empty list to avoid crash

def save_report_txt(report, filename="budget_report.txt"):
    # Saves the budget report as a plain text file (File I/O)
    try:
        with open(filename, "w") as f: # open file for writing
            f.write(report) # write the report string to the file
            print("Report saved to budget_report.txt") # confirm successful save
    except: # catch any error if file cannot be saved
        print("Error saving report.") # print error message to user

# VIEW HISTORY
def view_history(history):
    # Displays all previous budget sessions loaded from file.
    # Demonstrates: for loop, f-strings, list of dictionaries.
    if not history: # if list is empty = no saved sessions
        print("\n No previous sessions found.")
        return # exit the function early
    print(" FINANCIAL HISTORY") # print section title
    print("-" * 50) # print divider line
    i = 1 # start session counter at 1
    for session in history: # go through each saved session
        print(f"\n Session {i} - {session['name']}") # print session header
        print(f"  {session['date']}") # access nested dictionary
        print(f"  Total income:    ${session['income']['total']:.2f}") # access nested dictionary
        print(f"  Total expenses:  ${session['expenses']['total']:.2f}") # access nested dictionary
        print(f"  Total savings:   ${session['savings']['total']:.2f}") # access nested dictionary
        print(f"  Net balance:     ${session['summary']['net_balance']:.2f}") # access nested dictionary
        i = i + 1 # increase counter by 1

#VIEW SESSION DATA
def view_session_data(history):
    import pandas as pd 
    if not history: # if list is empty = no saved sessions
        print("\n No previous sessions found.")
        return # exit the function early
         
    print("SESSION DATA") # print section title
    print("-" * 50) # print divider line
    selection = int(input("Which session would you like to view? (enter the first as 1, second as 2, third as 3, and so on):"))
            
    while selection not in  range(1, (len(history) +1)): #this loops the user selection until the user chooses a session that exists in the history list
        print("This session does not exist. Please select a session that has occured:")
        selection = int(input("Which session would you like to view? (enter the first as 1, second as 2, third as 3, and so on):"))
            
    if selection in range(1, (len(history) +1)):
        index_selection = selection - 1 #turns the selected session into the correct index for the history list
        selected_session = history[index_selection] #extracts the session based on the user's choice
    
        print(f"\n Session {selection} - {selected_session['name']}") # print session header
        print(f"  {selected_session['date']}") # access nested dictionary
        income_data = {"Types": ["Monthly", "Other", "Total"], "Income": [selected_session['income']['monthly'],selected_session['income']['other'], selected_session['income']['total']]} # access nested dictionary
        income_df = pd.DataFrame(income_data)
        income_pivot = pd.pivot_table(income_df, values="Income", index="Types")
        display(income_pivot)
    
        expenses_data = {"Types": ["Food", "Transportation", "Entertainment", "Housing","Utilities", "Other", "Total"], 
                         "Expenses": [selected_session['expenses']['food'], selected_session['expenses']['transportation'], selected_session['expenses']['entertainment'],
                                      selected_session['expenses']['housing'], selected_session['expenses']['utilities'], selected_session['expenses']['other'],
                                      selected_session['expenses']['total']]} # access nested dictionary
        expenses_df = pd.DataFrame(expenses_data)
        expenses_pivot = pd.pivot_table(expenses_df, values="Expenses", index="Types")
        display(expenses_pivot)
    
        savings_data = {"Types": ["Savings", "Investments", "Retirement", "Total"], 
                            "Savings": [selected_session['savings']['savings'], selected_session['savings']['investments'], selected_session['savings']['retirement'],
                                        selected_session['savings']['total']]} # access nested dictionary
        savings_df = pd.DataFrame(savings_data)
        savings_pivot = pd.pivot_table(savings_df, values="Savings", index="Types")
        display(savings_pivot)
                
        summary_data = {"Types": ["Net Balance", "Expense Ratio"], 
                                 "Summary": [selected_session['summary']['net_balance'], selected_session['summary']['expense_ratio']]} # access nested dictionary
        summary_df = pd.DataFrame(summary_data)
        summary_pivot = pd.pivot_table(summary_df, values="Summary", index="Types")
        display(summary_pivot)
        
#MAIN MENU
def main():
    # Main function - runs the program with a menu.
    # Demonstrates: while loop, conditional logic, user interaction.
    print("\n" + "-" * 50) # print divider line
    print("  Welcome to the Budget Tracker! ") # welcome message
    print("-" * 50) # print divider line

    history = load_data() # load saved session from file

    while True: # keep showing menu until user chooses Exit
        print("\n Menu")
        print("1. Enter new budget")
        print("2. View history")
        print("3. View session data")
        print("4. Exit")

        choice = input("\nChoose an option (1/2/3/4): ").strip() # get user's menu choice

        if choice == "1":
            name = input("Enter your name: ").strip() # ask for user's name
            if not name: # if user pressed Enter without typing
                name = "User" # use default name
            tracker = BudgetTracker(name) # create new BudgetTracker object
            tracker.enter_income() # call method to collect income
            tracker.enter_expenses() # call method to collect expenses
            tracker.enter_savings() # call method to collect savings

            report = tracker.generate_report() # generate the budget report
            print(report) # print report to screen

            history.append(tracker.to_dict()) # add session to history list
            save_data(history) # save updated history to JSON file
            save_report_txt(report) # save report as text file
        elif choice == "2":
            view_history(history) # call function to display all saved sessions
        elif choice == "3":
            view_session_data(history) # call function to display a selected session's data
        elif choice == "4":
            print("\nGoodbye! ")
            break # exit the while True loop
            
        else:
            print("Invalid choice. Please enter 1, 2, 3, or 4.")
main()
```
